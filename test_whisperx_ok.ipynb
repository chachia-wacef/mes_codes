```python
import whisperx
import gc 
import torch
from whisperx.diarize import DiarizationPipeline
```

```python
device = "cuda" 
audio_file = "/kaggle/input/testmp3/003.mp3"
batch_size = 1 
compute_type = "float32"
```

```python
model = whisperx.load_model("large-v2", device, compute_type=compute_type)
```

```python
type(model)
```

```python
audio_file = "exp1.mp3"
```

```python
audio = whisperx.load_audio(audio_file)
```

```python
result = model.transcribe(audio, batch_size=batch_size)
```

```python
print(result["segments"])
```

```python
model_a, metadata = whisperx.load_align_model(language_code=result["language"], device=device)
result = whisperx.align(result["segments"], model_a, metadata, audio, device, return_char_alignments=False)

print(result["segments"])
```

```python
diarize_model = DiarizationPipeline(token='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', device=device)
```

```python
diarize_segments = diarize_model(audio)
```

```python
result = whisperx.assign_word_speakers(diarize_segments, result)
print(diarize_segments)
print(result["segments"])
```

```python
words = result["segments"]

source = []
for word in words:
    source = source + word["words"]

data = []
str1 = ""
last_speaker = None

for item in source:
    current_speaker = item.get("speaker")
    
    if current_speaker is not None:
        if last_speaker is None:
            last_speaker = "SPEAKER_01" 
        
        if current_speaker == last_speaker:
            str1 = str1 + " " +  item.get('word')
        else:
            data.append(str1)
            str1 = str(current_speaker) + ": "
            str1 = str1 + " " + item.get("word")
            last_speaker = current_speaker
    else:
        if last_speaker is None:
            last_speaker = "SPEAKER_01"
        
        str1 = str1 + " " + item.get('word')

else:
    data.append(str1)

for item in data:
    print(item)
```
