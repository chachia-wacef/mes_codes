```python
LOCAL_AUDIO = "exp1.mp3"
WAV_PATH = "exp1_16K.wav"

#décode le fichier audio (MP3, WAV, FLAC…) en tableau numpy
audio, sr = librosa.load(LOCAL_AUDIO, 
                         sr=16000, #Les modèles speech modernes sont tous entraînés en 16 kHz
                         #par exemple pour 8 kHz => trop pauvre, hautes fréquences perdues
                         mono=True) #Force 1 seul canal (mixage si stéréo) | NeMo n'accepte que mono en entrée → conversion obligatoire de toute façon

sf.write(WAV_PATH, audio, 16000, subtype="PCM_16") #PCM_16 : Définit la profondeur de bits 
print(f"Audio préparé : {WAV_PATH} | durée = {len(audio)/16000:.1f}s | SR = 16000 Hz")
```

+ Chunking manuel

```python
import soundfile as sf
import os

CHUNK_SEC = 30
OVERLAP_SEC = 2 #recouvrement entre 2 chunks consécutifs
SR = 16000

audio, sr = sf.read(WAV_PATH)
audio_duration = len(audio) / SR
print(f"Durée audio : {audio_duration:.1f}s")

chunk_samples = CHUNK_SEC * SR
overlap_samples = OVERLAP_SEC * SR
step = chunk_samples - overlap_samples

chunks_dir = "/tmp/chunks"
os.makedirs(chunks_dir, exist_ok=True)
for f in os.listdir(chunks_dir):
    os.remove(os.path.join(chunks_dir, f))

#Découper l'audio
chunk_files, chunk_offsets = [], []
start, i = 0, 0
while start < len(audio):
    end = min(start + chunk_samples, len(audio))
    path = f"{chunks_dir}/chunk_{i:04d}.wav"
    sf.write(path, audio[start:end], SR, subtype="PCM_16")
    chunk_files.append(path)
    chunk_offsets.append(start / SR)
    i += 1
    if end >= len(audio):
        break
    start += step

print(f"✓ {len(chunk_files)} chunks créés")
```

```python
all_predictions = canary.transcribe(
    chunk_files,
    batch_size=4, #4 chunks en parallèle
    pnc="yes", #Punctuation & Capitalization
    source_lang="fr",
    target_lang="fr",
    task="asr", #ASR (transcription)
    timestamps=True #Active les timestamps mot-à-mot
)
print(f"✓ {len(all_predictions)} chunks transcrits")
```

```python
full_text_parts = []
all_words = []

for chunk_idx, (pred, offset) in enumerate(zip(all_predictions, chunk_offsets)):
    text = pred.text if hasattr(pred, 'text') else str(pred)
    full_text_parts.append(text.strip())
    
    has_ts = hasattr(pred, 'timestamp') and pred.timestamp and pred.timestamp.get('word')
    
    if has_ts:
        ts_words = pred.timestamp['word']
        for w in ts_words:
            all_words.append({
                'word': w['word'],
                'start': w['start'] + offset, #timestamps absolus sur l'audio complet
                'end': w['end'] + offset #timestamps absolus sur l'audio complet
            })
        
        # Si quelques mots manquent (ex chunk 11 : 71 vs 70), on interpole les manquants à la fin
        text_words = text.strip().split()
        if len(text_words) > len(ts_words):
            missing = text_words[len(ts_words):]
            last_end = ts_words[-1]['end'] if ts_words else 0
            chunk_audio, _ = sf.read(chunk_files[chunk_idx])
            chunk_dur = len(chunk_audio) / SR
            remaining = max(0.5, chunk_dur - last_end)
            word_dur = remaining / len(missing)
            for j, word in enumerate(missing):
                all_words.append({
                    'word': word,
                    'start': offset + last_end + j * word_dur,
                    'end': offset + last_end + (j + 1) * word_dur,
                })
            print(f"  ℹ Chunk {chunk_idx} : {len(missing)} mot(s) interpolé(s) à la fin")
    
    else:
        # Cas chunk 0 : tout le texte sans timestamps → interpolation complète
        # 🐛 Le bug Canary
        # Parfois Canary te donne le texte mais "oublie" les timestamps (cas du chunk 0 chez toi : 60 mots, 0 timestamp).

        # → On invente les timestamps en répartissant les mots uniformément sur la durée du chunk.
        
        # Ex : chunk 0 dure 30s, contient 60 mots → on suppose 1 mot toutes les 0.5s :
        words_in_chunk = text.strip().split()
        if not words_in_chunk:
            continue
        chunk_audio, _ = sf.read(chunk_files[chunk_idx])
        chunk_dur = len(chunk_audio) / SR
        word_dur = chunk_dur / len(words_in_chunk)
        for j, word in enumerate(words_in_chunk):
            all_words.append({
                'word': word,
                'start': offset + j * word_dur,
                'end': offset + (j + 1) * word_dur,
            })
        print(f"  ⚠ Chunk {chunk_idx} : {len(words_in_chunk)} mots interpolés (aucun timestamp)")

full_text = " ".join(full_text_parts)

print()
print("=" * 60)
print("TRANSCRIPTION COMPLÈTE :")
print("=" * 60)
print(full_text)
print()
print(f"→ {len(all_words)} mots avec timestamps")
print(f"→ {len(full_text.split())} mots dans le texte")
print(f"→ Couverture : {all_words[-1]['end']:.1f}s / {audio_duration:.1f}s "
      f"({100*all_words[-1]['end']/audio_duration:.0f}%)")
```

```python
#Diarisation avec Sortformer

from nemo.collections.asr.models import SortformerEncLabelModel

# Chargement Sortformer (jusqu'à 4 locuteurs)
sortformer = SortformerEncLabelModel.from_pretrained("nvidia/diar_sortformer_4spk-v1")
sortformer.eval()
```

```python
def merge_short_segments(segments, gap_threshold=0.5):
    """Fusionne les segments du même speaker séparés par un petit gap."""
    if not segments:
        return segments
    merged = [segments[0].copy()]
    for seg in segments[1:]:
        last = merged[-1]
        if seg["speaker"] == last["speaker"] and seg["start"] - last["end"] < gap_threshold:
            last["end"] = seg["end"]
        else:
            merged.append(seg.copy())
    return merged
```

```python
# Diarisation sur l'audio complet (Sortformer gère les longs audios nativement)
diar_result = sortformer.diarize(audio=[WAV_PATH], batch_size=1, num_speakers=2)

# Parsing des segments
segments = []
print("=" * 60)
print("DIARISATION SORTFORMER :")
print("=" * 60)
for seg in diar_result[0]:
    parts = seg.split() #Chaque segment est une chaîne style : "12.34 17.89 speaker_0"
    seg_start, seg_end, spk = float(parts[0]), float(parts[1]), parts[2]
    segments.append({"start": seg_start, "end": seg_end, "speaker": spk})
    print(f"  [{seg_start:6.2f}s → {seg_end:6.2f}s] {spk}")

print(f"\n→ {len(segments)} segments | "
      f"locuteurs uniques : {set(s['speaker'] for s in segments)}")


# 👇nettoyage des segments avant fusion
segments = merge_short_segments(segments, gap_threshold=0.5)

print(f"→ {len(segments)} segments après merge")
```

```python
#Éliminer les "unknown" (assigner au speaker le plus proche)

def find_speaker(word_start, word_end, segments):
    """Assigne le speaker avec le plus grand overlap.
    Si aucun overlap (mot dans un gap), prend le segment le plus proche."""
    best_spk, best_overlap = None, 0
    #Pour chaque segment, calcule combien de temps le mot et le segment se chevauchent
    for seg in segments:
        overlap = max(0, min(word_end, seg["end"]) - max(word_start, seg["start"]))
        if overlap > best_overlap:
            best_overlap = overlap
            best_spk = seg["speaker"]
    # => Passe 1 = idéal (overlap réel = forte confiance)
    
    # Pas d'overlap → segment le plus proche temporellement
    #   Si le mot tombe dans un gap (silence, transition entre 2 segments), aucun segment ne le couvre.
    # → On prend le segment temporellement le plus proche.
    if best_spk is None:
        min_dist = float('inf')
        for seg in segments:
            # Distance au segment (0 si le mot est dans le segment)
            if word_end < seg["start"]:
                dist = seg["start"] - word_end
            elif word_start > seg["end"]:
                dist = word_start - seg["end"]
            else:
                dist = 0
            if dist < min_dist:
                min_dist = dist
                best_spk = seg["speaker"]

    # Passe 2 = sauvetage (élimine les "unknown" que tu avais — 88 mots chez toi)
    
    return best_spk if best_spk else "unknown"
#=> Cette passe garantit qu'aucun mot ne reste "unknown" : même dans un silence, il est rattaché au speaker du segment le plus proche temporellement.
```

```python
# Regroupe les mots de Canary par speaker en tours de parole
turns = []
current = None

for w in all_words:
    spk = find_speaker(w['start'], w['end'], segments)
    if current is None or current["speaker"] != spk:
        if current:
            turns.append(current)
        current = {
            "speaker": spk,
            "start": w['start'],
            "end": w['end'],
            "text": w['word']
        }
    else:
        current["end"] = w['end']
        current["text"] += " " + w['word']

if current:
    turns.append(current)

print("=" * 60)
print("RÉSULTAT FINAL (transcription + diarisation) :")
print("=" * 60)
for t in turns:
    print(f"\n[{t['start']:6.2f}s → {t['end']:6.2f}s] {t['speaker']}")
    print(f"  {t['text']}")

print(f"\n→ {len(turns)} tours de parole")
```
